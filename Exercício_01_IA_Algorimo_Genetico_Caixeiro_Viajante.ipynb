```python
import random
import numpy as np

# === MATRIZ DE DISTÂNCIAS ===

# Matriz que representa as distâncias entre as cidades - matriz_dist[i][j] é a distância da cidade i para a cidade j
matriz_dist = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

# Definições e inicialização de variáveis
num_cidades = len(matriz_dist)
populacao_tam = 50
geracoes = 300
taxa_mutacao = 0.1

# === Funções do Algoritmo Genético ===

# Função para avaliar a qualidade de um caminho - Quanto menor a distância total, maior o fitness (1/distância)
def fitness(caminho):
    dist = 0
    for i in range(len(caminho) - 1):
        dist += matriz_dist[caminho[i]][caminho[i+1]]
    dist += matriz_dist[caminho[-1]][caminho[0]]
    return 1 / dist

# Função para cria um indivíduo (solução) inicial aleatório - Um indivíduo é uma permutação das cidades, representando uma possível rota.
def criar_individuo():
    caminho = list(range(num_cidades))
    random.shuffle(caminho)
    return caminho

# Função recombinação: combina dois pais para criar um filho.
def crossover(pai1, pai2):
    ini, fim = sorted(random.sample(range(num_cidades), 2))
    filho = [None] * num_cidades
    filho[ini:fim] = pai1[ini:fim]
    pos = fim # Preenche o restante do filho com as cidades do pai2 na ordem em que aparecem, pulando as cidades que já foram incluídas do pai1.
    for cidade in pai2:
        if cidade not in filho:
            if pos >= num_cidades:
                pos = 0
            filho[pos] = cidade
            pos += 1
    return filho

# Função que introduz pequenas alterações aleatórias em um indivíduo - Troca a posição de duas cidades no caminho com uma certa probabilidade.
def mutacao(individuo):
    if random.random() < taxa_mutacao:
        i, j = random.sample(range(num_cidades), 2)
        individuo[i], individuo[j] = individuo[j], individuo[i]
    return individuo

# === Algoritmo Genético ===

# Inicializa a população com indivíduos aleatórios.
populacao = [criar_individuo() for _ in range(populacao_tam)]

# Loop principal do algoritmo genético, executando por um número definido de gerações.
for g in range(geracoes):
    populacao = sorted(populacao, key=lambda ind: -fitness(ind))  # Avalia a população e a ordena pelo fitness (do melhor para o pior).
    nova_pop = populacao[:10]  # Cria a nova população, começando com os melhores indivíduos da geração atual (elitismo).
    while len(nova_pop) < populacao_tam: # Gera novos indivíduos através de crossover e mutação até atingir o tamanho da população.
        pai1, pai2 = random.sample(populacao[:25], 2) # Seleciona dois pais para crossover (geralmente entre os melhores).
        filho = crossover(pai1, pai2) # Realiza o crossover para criar um filho.
        filho = mutacao(filho)  # Aplica mutação ao filho.
        nova_pop.append(filho) # Adiciona o filho à nova população.
    populacao = nova_pop # A nova população substitui a antiga.

# Após todas as gerações, encontra o melhor caminho e calcula a distância total percorrida nele
melhor = max(populacao, key=fitness)
melhor_dist = 1 / fitness(melhor)

# Exibe o melhor caminho e a distância total percorrida nele.
print("Melhor caminho encontrado:", melhor)
print("Distância total:", melhor_dist)
```
