```python
import random
import numpy as np

# === MATRIZ DE DISTÂNCIAS ===
matriz_dist = [

#    A  B  C  D  E
    [0, 6, 2, 2, 3],  # A
    [6, 0, 3, 4, 3],  # B
    [2, 3, 0, 3, 6],  # C
    [2, 4, 3, 0, 5],  # D
    [3, 3, 6, 5, 0]   # E
]

# Lista com nomes das cidades
cidades = ["A", "B", "C", "D", "E"]

# Definições e inicialização de variáveis
num_cidades = len(matriz_dist)
populacao_tam = 50
geracoes = 300
taxa_mutacao = 0.1

# === Funções do Algoritmo Genético ===
def fitness(caminho):
    dist = 0
    for i in range(len(caminho) - 1):
        dist += matriz_dist[caminho[i]][caminho[i+1]]
    dist += matriz_dist[caminho[-1]][caminho[0]]  # volta ao início
    return 1 / dist

def criar_individuo():
    caminho = list(range(num_cidades))
    random.shuffle(caminho)
    return caminho

def crossover(pai1, pai2):
    ini, fim = sorted(random.sample(range(num_cidades), 2))
    filho = [None] * num_cidades
    filho[ini:fim] = pai1[ini:fim]
    pos = fim
    for cidade in pai2:
        if cidade not in filho:
            if pos >= num_cidades:
                pos = 0
            filho[pos] = cidade
            pos += 1
    return filho

def mutacao(individuo):
    if random.random() < taxa_mutacao:
        i, j = random.sample(range(num_cidades), 2)
        individuo[i], individuo[j] = individuo[j], individuo[i]
    return individuo

# === Algoritmo Genético ===
populacao = [criar_individuo() for _ in range(populacao_tam)]

for g in range(geracoes):
    populacao = sorted(populacao, key=lambda ind: -fitness(ind))
    nova_pop = populacao[:10]
    while len(nova_pop) < populacao_tam:
        pai1, pai2 = random.sample(populacao[:25], 2)
        filho = crossover(pai1, pai2)
        filho = mutacao(filho)
        nova_pop.append(filho)
    populacao = nova_pop

melhor = max(populacao, key=fitness)
melhor_dist = 1 / fitness(melhor)
melhor_caminho_nomes = [cidades[i] for i in melhor]

# Exibe o resultado
print("Melhor caminho encontrado:", " -> ".join(melhor_caminho_nomes + [melhor_caminho_nomes[0]]))
print("Distância total:", melhor_dist)
```
